### Funciones auxiliares

- `sha256()` calcula la huella del contenido de cada archivo. Decidimos detectar duplicados por contenido y no por nombre, porque una misma foto puede tener nombres distintos (`IMG_0530.HEIC` e `IMG_0530(1).HEIC`) y dos fotos distintas pueden llamarse igual, ya que cada teléfono numera sus fotos por su cuenta.
- `metadatos()` lee de cada foto el tamaño, la cámara, la fecha, la orientación y si trae mapa de profundidad. Solo lee el encabezado del archivo, sin decodificar los píxeles, para que sea rápido.
- `norm()` pasa un nombre de columna a minúsculas y sin acentos ni símbolos ("ID - VACA" → "idvaca"), para encontrar las columnas aunque cambien los espacios o los acentos.
- `texto()` convierte una celda de Excel a texto (1221.0 → "1221", y una celda con formato de hora → "HH MM").
- `clave()` y `ordenar_ids()` ordenan los IDs como números, para que 1204 quede antes de 91549.

```python
def sha256(ruta, bloque=1 << 20):
    """Huella del contenido del archivo: dos archivos con la misma huella son idénticos."""
    h = hashlib.sha256()
    with open(ruta, "rb") as f:
        for b in iter(lambda: f.read(bloque), b""):
            h.update(b)
    return h.hexdigest()


def metadatos(ruta):
    """Tamaño, cámara, fecha, orientación y profundidad de una foto (sin decodificar los píxeles)."""
    r = {"ancho": None, "alto": None, "camara": "", "fecha": "", "orientacion": None,
         "profundidad": False, "error": ""}
    try:
        with Image.open(ruta) as im:
            exif = im.getexif()
            r["ancho"], r["alto"] = im.size
            r["camara"] = f"{exif.get(0x010F, '')} {exif.get(0x0110, '')}".strip()
            fecha = exif.get_ifd(0x8769).get(0x9003) or exif.get(0x0132) or ""
            r["fecha"] = str(fecha)[:10].replace(":", "-")
            r["orientacion"] = exif.get(0x0112)
            r["profundidad"] = bool(im.info.get("depth_images"))
    except Exception as e:
        r["error"] = str(e)[:120]
    return r


def norm(texto):
    """Nombre de columna en minúsculas, sin acentos ni símbolos ('ID - VACA' -> 'idvaca')."""
    t = str(texto).lower().translate(str.maketrans("áéíóúüñ", "aeiouun"))
    return re.sub(r"[^a-z0-9]", "", t)


def texto(v):
    """Celda de Excel a texto (1221.0 -> '1221', una hora -> 'HH MM')."""
    if v is None:
        return ""
    if isinstance(v, float) and v.is_integer():
        return str(int(v))
    if isinstance(v, (dt.time, dt.datetime)):
        return v.strftime("%H %M")
    return str(v).strip()


def clave(x):
    """Ordena los IDs como números cuando lo son."""
    return (0, int(x)) if str(x).isdigit() else (1, str(x))


def ordenar_ids(ids):
    return sorted(ids, key=clave)
```

## 2.1 Recolección de los datos

### Inventario de las fuentes

Recorremos las carpetas de fotos y, por cada imagen, guardamos su vaca, la descarga de la que viene, su visita, su tamaño, su huella y sus metadatos.

Tomamos estas decisiones.

- La vaca de cada foto es el nombre de la carpeta que la contiene, porque así estaban organizadas en Drive (una carpeta por vaca).
- La visita de cada foto sale del nombre de su carpeta de primer nivel, ya que las carpetas de la visita del 14/10 traen el sufijo de descarga de Drive y las de la primera visita (29/09) no. En 2.2 comprobamos que coincida con la fecha de captura de los metadatos.
- Solo contamos archivos de imagen (HEIC, HEIF, JPG, JPEG y PNG) e ignoramos los archivos ocultos del sistema.
- Las vacas de ordeño se toman del nombre de cada CSV, porque cada archivo se llama como el ID de su vaca, y aquí no se lee su contenido.

```python
EXT_FOTO = {".heic", ".heif", ".jpg", ".jpeg", ".png"}
filas = []
for p in sorted(DIR_FOTOS.rglob("*")):
    if p.is_file() and p.suffix.lower() in EXT_FOTO and not p.name.startswith("."):
        rel = p.relative_to(DIR_FOTOS)
        filas.append({"id_vaca": p.parent.name.strip(), "descarga": rel.parts[0],
                      "visita": "14/10" if PATRON_DESCARGA.search(rel.parts[0]) else "29/09",
                      "archivo": p.name, "ruta": str(rel), "ext": p.suffix.lower(),
                      "bytes": p.stat().st_size, "sha256": sha256(p)})
fotos = pd.DataFrame(filas)
fotos = pd.concat([fotos, pd.DataFrame([metadatos(DIR_FOTOS / r) for r in fotos["ruta"]])], axis=1)

csv_ordeno = sorted(p for p in DIR_CSV.rglob("*.csv") if p.stem.strip().isdigit())
ids_ordeno = {p.stem.strip() for p in csv_ordeno}

fuentes = pd.DataFrame([
    {"fuente": "Fotos de la visita del 14/10 (fotosextraidas/)", "archivos": int((fotos["visita"] == "14/10").sum()),
     "MB": fotos.loc[fotos["visita"] == "14/10", "bytes"].sum() / 1e6},
    {"fuente": "Fotos de la visita del 29/09 (fotosextraidas/)", "archivos": int((fotos["visita"] == "29/09").sum()),
     "MB": fotos.loc[fotos["visita"] == "29/09", "bytes"].sum() / 1e6},
    {"fuente": "Bitácora de expertos del 14/10 (etiquetas_bsc.xlsx)", "archivos": 1, "MB": BITACORA.stat().st_size / 1e6},
    {"fuente": "Hoja manuscrita del 29/09 (bitacora_hoja_manuscrita.csv)", "archivos": 1, "MB": HOJA.stat().st_size / 1e6},
    {"fuente": "Dataset VMS 3 (vms-3.csv)", "archivos": 1, "MB": VMS3.stat().st_size / 1e6},
    {"fuente": "CSV de ordeño (uno por vaca)", "archivos": len(csv_ordeno),
     "MB": sum(p.stat().st_size for p in csv_ordeno) / 1e6},
]).round({"MB": 2})
display(fuentes)
```

### Bitácora de expertos

La bitácora es un Excel compartido y sus encabezados no están en la primera fila. En lugar de suponer una fila fija, buscamos en las primeras 40 filas de cada hoja la que tenga columnas que empiecen con "ID" y "BCS". Guardamos el número real de fila de Excel (`fila_excel`) para poder ubicar cada evaluación en el archivo original.

Tomamos estas decisiones.

- Usamos solo el ID, la hora y el BCS. Las columnas "Foto" y "Carpeta" son ligas de Drive y no aportan datos para el análisis.
- El BCS se convierte a número aceptando punto o coma decimal. Si una celda no es número, queda vacía y se reporta en 2.4.

```python
# Bitácora: se busca la fila que tiene los encabezados "ID" y "BCS" (puede no ser la primera)
wb = load_workbook(BITACORA, read_only=True, data_only=True)
hoja, fila_enc = None, None
for ws in wb.worksheets:
    for i, fila in enumerate(ws.iter_rows(min_row=1, max_row=40, values_only=True), start=1):
        nombres = [norm(v) for v in fila if v not in (None, "")]
        if any(n.startswith("id") for n in nombres) and any(n.startswith("bcs") for n in nombres):
            hoja, fila_enc = ws, i
            break
    if hoja is not None:
        break
nombre_hoja = hoja.title
filas = list(hoja.iter_rows(min_row=fila_enc, values_only=True))
wb.close()

encabezado = [texto(c) or f"(columna {i + 1})" for i, c in enumerate(filas[0])]
datos = []
for num, fila in enumerate(filas[1:], start=fila_enc + 1):
    valores = [texto(c) for c in fila][:len(encabezado)]
    if any(valores):
        datos.append([num] + valores + [""] * (len(encabezado) - len(valores)))
bit_raw = pd.DataFrame(datos, columns=["fila_excel"] + encabezado)


def columna(df, *prefijos):
    for c in df.columns:
        if c != "fila_excel" and any(norm(c).startswith(p) for p in prefijos):
            return c
    raise KeyError(f"No encontré una columna que empiece con {prefijos}")


c_id, c_hora, c_bcs = columna(bit_raw, "idvaca", "id"), columna(bit_raw, "hora"), columna(bit_raw, "bcs")
bit = pd.DataFrame({
    "fila_excel": bit_raw["fila_excel"],
    "id_vaca": bit_raw[c_id].str.replace(r"\.0$", "", regex=True),
    "hora_original": bit_raw[c_hora],
    "bcs": pd.to_numeric(bit_raw[c_bcs].str.replace(",", "."), errors="coerce"),
})
print(f"Bitácora: hoja '{nombre_hoja}', encabezados en la fila {fila_enc}, {len(bit)} filas")
print("Columnas:", [c for c in bit_raw.columns if c != "fila_excel"])
```

### Dataset VMS 3

No sabemos con qué programa se exportó `vms-3.csv`, así que probamos primero la codificación UTF-8 y después Latin-1, y dejamos que pandas detecte el separador. Leemos todo como texto para no perder formatos.

El ID de la vaca está en la columna "Número del animal". No usamos "Número(s) de selección de animal", que es otra cosa. Si la columna del ID no aparece, se toma la primera.

```python
# Dataset VMS 3: se prueba la codificación y se detecta el separador
def leer_csv(ruta):
    for enc in ("utf-8-sig", "latin-1"):
        try:
            return pd.read_csv(ruta, sep=None, engine="python", encoding=enc,
                               dtype=str, keep_default_na=False)
        except UnicodeDecodeError:
            continue
    raise ValueError(f"No pude leer {ruta}")


vms = leer_csv(VMS3)
vms.columns = [str(c).strip() for c in vms.columns]
columnas_vms = list(vms.columns)
c_vms_id = next((c for c in columnas_vms if norm(c).startswith("numerodelanimal")), columnas_vms[0])
c_evento = next((c for c in columnas_vms if "eventobcs" in norm(c)), None)
vms["id_vaca"] = vms[c_vms_id].str.strip().str.replace(r"\.0$", "", regex=True)
print(f"vms-3.csv: {len(vms)} filas y {len(columnas_vms)} columnas")
print(f"ID de la vaca: '{c_vms_id}' | Evento BCS: '{c_evento}'")
print(f"CSV de ordeño: {len(csv_ordeno)} vacas")
```

## 2.2 Descripción de los datos

### Bitácora de expertos

Resumimos cuántas evaluaciones y vacas tiene la bitácora, y la estadística básica del BCS.

La hora se clasifica en tres formatos. "HH MM" es el esperado (por ejemplo "13 00"), "x" indica que no se registró y "otro formato" agrupa cualquier otra cosa.

```python
hora = bit["hora_original"].str.strip()
formato_hora = pd.Series(np.select(
    [hora.str.lower().eq("x"), hora.str.match(r"^\d{1,2}\s\d{2}$")],
    ["x (sin dato)", "HH MM"], "otro formato"), index=bit.index)

resumen_bit = pd.Series({
    "evaluaciones (filas)": len(bit),
    "vacas distintas": bit["id_vaca"].nunique(),
    "vacas con más de una evaluación": int((bit["id_vaca"].value_counts() > 1).sum()),
    "BCS mínimo": bit["bcs"].min(),
    "BCS máximo": bit["bcs"].max(),
    "BCS promedio": round(bit["bcs"].mean(), 3),
    "BCS mediana": bit["bcs"].median(),
    "BCS desviación estándar": round(bit["bcs"].std(), 3),
    "todos en pasos de 0.25": bool(((bit["bcs"].dropna() * 4) % 1 == 0).all()),
}, dtype=object)
display(resumen_bit.to_frame("valor"))
display(formato_hora.value_counts().rename_axis("formato de la hora").to_frame("filas"))
```

### Hoja manuscrita

Resumimos la hoja igual que la bitácora. La hoja no registra la hora. También contamos cuántas vacas de la hoja están en la bitácora del 14/10, porque esas vacas tienen BCS de las dos visitas.

Al final juntamos las dos fuentes, que son todas las evaluaciones de BCS que tenemos.

```python
def resumen_bcs(df):
    b, n = df["bcs"], df["id_vaca"].value_counts()
    return {
        "evaluaciones": len(df),
        "vacas distintas": df["id_vaca"].nunique(),
        "vacas con más de una evaluación": int((n > 1).sum()),
        "vacas con tres evaluaciones o más": int((n > 2).sum()),
        "BCS mínimo": b.min(),
        "BCS máximo": b.max(),
        "BCS promedio": round(b.mean(), 3),
        "BCS mediana": b.median(),
        "BCS desviación estándar": round(b.std(), 3),
        "todos en pasos de 0.25": bool(((b.dropna() * 4) % 1 == 0).all()),
    }


ids_bit_14, ids_hoja_29 = set(bit["id_vaca"]), set(hoja["id_vaca"])
evals = pd.concat([bit[["id_vaca", "bcs"]].assign(fuente="bitácora 14/10"),
                   hoja[["id_vaca", "bcs"]].assign(fuente="hoja 29/09")], ignore_index=True)
display(pd.DataFrame({"hoja 29/09": resumen_bcs(hoja), "bitácora + hoja": resumen_bcs(evals)}, dtype=object))
print(f"Vacas de la hoja también en la bitácora del 14/10: {len(ids_hoja_29 & ids_bit_14)} | "
      f"solo en la hoja: {len(ids_hoja_29 - ids_bit_14)} {ordenar_ids(ids_hoja_29 - ids_bit_14)}")
fresca = hoja[hoja["nota"].str.lower() == "fresca"]
print(f"Renglones con la nota 'Fresca': {len(fresca)} {ordenar_ids(set(fresca['id_vaca']))}")
```

### Fotografías

Una foto "única" es un contenido distinto. Si dos archivos tienen la misma huella cuentan como una sola foto, aunque estén en carpetas distintas o tengan otro nombre, y por eso reportamos el total con copias y sin copias.

Separamos las cifras por visita. La cámara, la resolución y la fecha salen de los metadatos EXIF de cada foto, y la tabla de fechas sirve para comprobar que la visita que sale del nombre de la carpeta coincide con la fecha de captura. "Mapa de profundidad" indica si el HEIC trae la información del sensor LiDAR del iPhone.

```python
unicas = fotos.drop_duplicates("sha256")


def resumen_de(df):
    u = df.drop_duplicates("sha256")
    return {
        "archivos": len(df),
        "fotos únicas": len(u),
        "tamaño total (GB)": round(df["bytes"].sum() / 1e9, 2),
        "tamaño sin copias (GB)": round(u["bytes"].sum() / 1e9, 2),
        "carpetas descargadas": df[["id_vaca", "descarga"]].drop_duplicates().shape[0],
        "vacas (IDs de carpeta)": df["id_vaca"].nunique(),
        "fotos con mapa de profundidad": int(u["profundidad"].sum()),
        "fotos que no se pudieron leer": int((u["error"] != "").sum()),
    }


def por_visita(serie, nombre):
    # Fotos únicas por categoría y visita, ordenadas de mayor a menor, con el total al final
    t = pd.crosstab(serie, unicas["visita"]).reindex(columns=["14/10", "29/09"], fill_value=0)
    t["total"] = t.sum(axis=1)
    t = t.sort_values("total", ascending=False)
    t.loc["total"] = t.sum()
    t.index.name = nombre
    return t


resumen_fotos = pd.DataFrame({v: pd.Series(resumen_de(df), dtype=object) for v, df in
                              (("14/10", fotos[fotos["visita"] == "14/10"]),
                               ("29/09", fotos[fotos["visita"] == "29/09"]), ("total", fotos))})
display(resumen_fotos)
display(por_visita(unicas["ext"], "formato"))
display(por_visita(unicas["camara"].replace("", "(sin dato)"), "cámara"))
resolucion = unicas["ancho"].astype("Int64").astype(str) + " × " + unicas["alto"].astype("Int64").astype(str)
display(por_visita(resolucion, "resolución (px)"))
display(por_visita(unicas["fecha"].replace("", "(sin dato)"), "fecha de captura"))
```

### Fotos por vaca

```python
pv = fotos.drop_duplicates(["id_vaca", "sha256"])
por_vaca_visita = pd.crosstab(pv["id_vaca"], pv["visita"]).reindex(columns=["14/10", "29/09"], fill_value=0)
por_vaca_visita["total"] = por_vaca_visita.sum(axis=1)
por_vaca_visita = por_vaca_visita.loc[ordenar_ids(por_vaca_visita.index)]
por_vaca = por_vaca_visita["total"].rename("fotos únicas")
display(por_vaca.describe().round(2).to_frame("fotos únicas por vaca"))
display(por_vaca_visita)
```

## 2.3 Exploración de los datos

### Distribución del BCS

Tomamos estas decisiones.

- Contamos evaluaciones y no vacas, así que una vaca evaluada dos veces aporta dos valores y el histograma refleja todos los BCS que dieron los expertos.
- Juntamos las dos fuentes (bitácora del 14/10 y hoja del 29/09) y mostramos cuántas evaluaciones aporta cada una.
- Mostramos todas las clases de 0.25 entre el mínimo y el máximo, aunque alguna tenga cero evaluaciones, para que se vean los huecos.

Esta distribución importa para las dos métricas del proyecto. Con pocas evaluaciones en una clase la Accuracy puede cambiar mucho con un solo error, y el MAE depende de qué tan lejos quedan las clases extremas del centro.

La gráfica se guarda en `figuras/distribucion_bcs.png`.

```python
FUENTES = ["bitácora 14/10", "hoja 29/09"]
COLORES = {"bitácora 14/10": "#2a78d6", "hoja 29/09": "#eb6834"}

bcs_validos = evals["bcs"].dropna()
clases = np.arange(np.floor(bcs_validos.min() * 4) / 4, bcs_validos.max() + 0.125, 0.25)
conteo = (evals.dropna(subset=["bcs"]).groupby("fuente")["bcs"].value_counts().unstack(0)
          .reindex(index=clases, columns=FUENTES).fillna(0).astype(int))
tabla_clases = conteo.copy()
tabla_clases["evaluaciones"] = conteo.sum(axis=1)
tabla_clases["porcentaje"] = (100 * tabla_clases["evaluaciones"] / tabla_clases["evaluaciones"].sum()).round(1)
tabla_clases.index = pd.Index([f"{c:.2f}" for c in clases], name="BCS")
display(tabla_clases)

fig, ax = plt.subplots(figsize=(8, 4))
base = np.zeros(len(tabla_clases))
for f in FUENTES:
    ax.bar(tabla_clases.index, tabla_clases[f], bottom=base, width=0.6, color=COLORES[f],
           edgecolor="white", linewidth=2, label=f)
    base += tabla_clases[f].values
for x, y in zip(tabla_clases.index, tabla_clases["evaluaciones"]):
    ax.text(x, y + 0.4, str(y), ha="center", fontsize=9, color="#333333")
ax.set_xlabel("BCS asignado por los expertos")
ax.set_ylabel("Número de evaluaciones")
ax.set_title(f"Distribución del BCS ({len(bcs_validos)} evaluaciones de {evals['id_vaca'].nunique()} vacas)")
ax.legend(frameon=False)
ax.grid(axis="y", color="#e5e5e5", linewidth=0.8)
ax.set_axisbelow(True)
ax.spines[["top", "right"]].set_visible(False)
fig.tight_layout()
fig.savefig(FIGURAS / "distribucion_bcs.png", dpi=200)
plt.show()
```

### Vacas con más de una evaluación

Hay dos casos y los medimos por separado.

- Algunas vacas aparecen dos veces en la bitácora del 14/10 o dos veces en la hoja del 29/09, porque las evaluaron expertos distintos el mismo día. Esa diferencia es la referencia de cuánto varía el BCS cuando lo asignan personas distintas.
- Otras vacas se evaluaron en las dos visitas, y en ellas comparamos el promedio de la hoja contra el promedio de la bitácora. Si la diferencia entre fechas es mayor que la diferencia entre expertos, no da igual con qué fecha se etiqueta una foto.

Las evaluaciones se ordenan según la fila de la bitácora o el renglón de la hoja. Ese orden no indica qué experto dio cada valor.

```python
def pares_de(df, orden):
    # Evaluaciones de las vacas que aparecen más de una vez, una columna por evaluación
    rep = df[df.duplicated("id_vaca", keep=False)].sort_values(["id_vaca", orden])
    rep = rep.assign(orden=rep.groupby("id_vaca").cumcount() + 1)
    p = rep.pivot(index="id_vaca", columns="orden", values="bcs")
    p.columns = [f"bcs_{c}" for c in p.columns]
    p["diferencia"] = p.max(axis=1) - p.min(axis=1)
    return p.loc[ordenar_ids(p.index)]


# Mismo día: bitácora del 14/10 y hoja del 29/09
pares = pares_de(bit, "fila_excel")
dif = pares["diferencia"]
display(pares)
print(f"Bitácora 14/10: {len(pares)} vacas con más de una evaluación | diferencia media {dif.mean():.3f}, "
      f"máxima {dif.max():.2f}, idénticas {(dif == 0).sum()}")
pares_hoja = pares_de(hoja, "renglon")
display(pares_hoja)
print(f"Hoja 29/09: {len(pares_hoja)} vacas con más de una evaluación | diferencia media "
      f"{pares_hoja['diferencia'].mean():.3f}, máxima {pares_hoja['diferencia'].max():.2f}")

# Fechas distintas: promedio de la hoja menos promedio de la bitácora
prom = evals.groupby(["id_vaca", "fuente"])["bcs"].mean().unstack().reindex(columns=FUENTES)
entre_fechas = prom.dropna().copy()
entre_fechas["hoja_menos_bitacora"] = entre_fechas["hoja 29/09"] - entre_fechas["bitácora 14/10"]
entre_fechas = entre_fechas.loc[ordenar_ids(entre_fechas.index)]
display(entre_fechas.round(3))
d_fechas = entre_fechas["hoja_menos_bitacora"].abs()
print(f"Vacas con BCS en las dos fechas: {len(entre_fechas)} | diferencia absoluta media {d_fechas.mean():.3f}, "
      f"máxima {d_fechas.max():.2f} | iguales {(d_fechas == 0).sum()} | mayores a 0.25: {(d_fechas > 0.25).sum()} "
      f"{ordenar_ids(entre_fechas.index[d_fechas > 0.25])}")
print(f"Diferencia con signo media (hoja - bitácora): {entre_fechas['hoja_menos_bitacora'].mean():+.3f}")
```

### Cruce entre fuentes

Comparamos los IDs de la bitácora del 14/10, la hoja del 29/09, las carpetas de fotos, `vms-3.csv` y los CSV de ordeño. Una vaca "tiene BCS" si está en la bitácora o en la hoja, y está "en ordeño" si existe un CSV con su ID. El cruce indica con cuántas vacas se pueden combinar las fotos y el BCS con las otras fuentes.

```python
ids_bit, ids_hoja, ids_fotos, ids_vms = set(bit["id_vaca"]), set(hoja["id_vaca"]), set(fotos["id_vaca"]), set(vms["id_vaca"])
ids_bcs = ids_bit | ids_hoja
conteos = pd.Series({
    "con BCS (bitácora u hoja)": len(ids_bcs),
    "en la bitácora del 14/10": len(ids_bit),
    "en la hoja del 29/09": len(ids_hoja),
    "en la bitácora y en la hoja": len(ids_bit & ids_hoja),
    "con BCS y fotos": len(ids_bcs & ids_fotos),
    "con BCS y vms-3": len(ids_bcs & ids_vms),
    "con BCS y ordeño": len(ids_bcs & ids_ordeno),
    "con BCS, vms-3 y ordeño": len(ids_bcs & ids_vms & ids_ordeno),
    "en vms-3 sin BCS de expertos": len(ids_vms - ids_bcs),
    "con fotos y sin BCS": len(ids_fotos - ids_bcs),
})
display(conteos.to_frame("vacas"))
print("Con BCS y en vms-3:", ordenar_ids(ids_bcs & ids_vms))
print("   de ellas, solo con BCS de la hoja:", ordenar_ids((ids_hoja - ids_bit) & ids_vms))
print("Con BCS y ordeño:", ordenar_ids(ids_bcs & ids_ordeno))
print("Con BCS, vms-3 y ordeño:", ordenar_ids(ids_bcs & ids_vms & ids_ordeno))
print("En vms-3 sin BCS de expertos:", ordenar_ids(ids_vms - ids_bcs))
print("Con fotos y sin BCS:", ordenar_ids(ids_fotos - ids_bcs))

cruce = pd.DataFrame({"id_vaca": ordenar_ids(ids_bcs | ids_fotos | ids_vms)})
for nombre, ids in (("bitácora", ids_bit), ("hoja", ids_hoja), ("fotos", ids_fotos), ("vms-3", ids_vms), ("ordeño", ids_ordeno)):
    cruce[nombre] = np.where(cruce["id_vaca"].isin(ids), "Sí", "No")
display(cruce)
```

### "Evento BCS" del dataset contra el BCS de los expertos

Todavía no sabemos qué mide la columna "Evento BCS" de `vms-3.csv`, así que la comparamos con el BCS de los expertos para ver si coinciden.

Tomamos estas decisiones.

- Primero la comparamos con la bitácora del 14/10, como en la versión anterior de este análisis, y después con la hoja, solo para las vacas que no están en la bitácora.
- Usamos el promedio de las evaluaciones de cada vaca, porque "Evento BCS" tiene un solo valor por vaca.
- La diferencia es Evento BCS menos el BCS de los expertos, así que es positiva cuando el dataset da un valor mayor.

```python
if c_evento is None:
    print("vms-3.csv no tiene una columna 'Evento BCS'")
else:
    evento = vms[["id_vaca", c_evento]].rename(columns={c_evento: "evento_bcs"})
    evento["evento_bcs"] = pd.to_numeric(evento["evento_bcs"].str.replace(",", "."), errors="coerce")
    evento = evento.dropna(subset=["evento_bcs"])

    expertos = bit.groupby("id_vaca")["bcs"].agg(bcs_expertos="mean", evaluaciones="size").reset_index()
    comp = expertos.merge(evento, on="id_vaca")
    comp["diferencia"] = (comp["evento_bcs"] - comp["bcs_expertos"]).round(2)
    print("Contra la bitácora del 14/10:")
    display(comp.sort_values("diferencia", ascending=False).reset_index(drop=True))
    print(f"{len(comp)} vacas con ambos valores: Evento BCS mayor en {(comp['diferencia'] > 0).sum()}, "
          f"igual en {(comp['diferencia'] == 0).sum()} y menor en {(comp['diferencia'] < 0).sum()}")

    solo_hoja = (hoja[~hoja["id_vaca"].isin(ids_bit)].groupby("id_vaca")["bcs"]
                 .agg(bcs_hoja="mean", evaluaciones="size").reset_index())
    comp_hoja = solo_hoja.merge(evento, on="id_vaca")
    comp_hoja["diferencia"] = (comp_hoja["evento_bcs"] - comp_hoja["bcs_hoja"]).round(2)
    print("\nContra la hoja del 29/09 (vacas que no están en la bitácora):")
    display(comp_hoja.sort_values("diferencia", ascending=False).reset_index(drop=True))
    print(f"{len(comp_hoja)} vacas con ambos valores: Evento BCS mayor en {(comp_hoja['diferencia'] > 0).sum()}, "
          f"igual en {(comp_hoja['diferencia'] == 0).sum()} y menor en {(comp_hoja['diferencia'] < 0).sum()}")
```

### Archivos repetidos

- Hay copias dentro de una vaca cuando dos archivos de esa vaca tienen la misma huella. Pasa cuando una carpeta se descargó dos veces o cuando hay archivos tipo `IMG_0530(1).HEIC`. En la primera visita muchos nombres con "(1)" son tomas distintas y no copias, por eso las copias se cuentan por huella y no por nombre.
- Si una vaca tiene más de una carpeta del 14/10 con las mismas fotos, es una descarga repetida, pero si las carpetas tienen fotos distintas no se puede saber qué fotos corresponden a cada evaluación. En la primera visita hay una carpeta por vaca, así que esta revisión solo aplica a las carpetas del 14/10.
- También contamos las fotos de la primera visita que repiten una foto del 14/10 de la misma vaca.

```python
# Copias exactas y descargas repetidas
pv_visita = fotos.drop_duplicates(["id_vaca", "sha256"]).groupby("visita").size()
copias_visita = (fotos.groupby("visita").size() - pv_visita).reindex(["14/10", "29/09"], fill_value=0)
copias = int(copias_visita.sum())
es_numerada = fotos["archivo"].str.contains(r"\(\d+\)\.", regex=True)
numeradas = int(es_numerada[fotos["visita"] == "14/10"].sum())
f14, f29 = fotos[fotos["visita"] == "14/10"], fotos[fotos["visita"] == "29/09"]
descargas = f14.groupby(["id_vaca", "descarga"])["sha256"].apply(frozenset)
dobles = {v: ("mismo contenido" if len(set(g.values)) == 1 else "contenido distinto")
          for v, g in descargas.groupby(level=0) if len(g) > 1}
sha_14 = set(zip(f14["id_vaca"], f14["sha256"]))
repiten_14 = sum((v, s) in sha_14 for v, s in set(zip(f29["id_vaca"], f29["sha256"])))
print(f"Archivos repetidos dentro de una misma vaca: {copias} "
      f"({copias_visita['14/10']} del 14/10 y {copias_visita['29/09']} del 29/09)")
print(f"Archivos con nombre tipo '(1)': {numeradas} del 14/10 y {int(es_numerada[fotos['visita'] == '29/09'].sum())} del 29/09")
print(f"Vacas con más de una carpeta del 14/10: {len(dobles)}")
display(pd.Series(dobles, dtype=object).rename_axis("id_vaca").to_frame("carpetas"))
print(f"Fotos del 29/09 que repiten una foto del 14/10 de la misma vaca: {repiten_14}")
```
